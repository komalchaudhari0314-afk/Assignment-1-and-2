# 1. Working with lists

# Create a list on integers

```python
numbers = [10, 20, 30, 40, 50]
print(numbers)
```

# Append a new number

```python
numbers = [10, 20, 30, 40]
numbers.append(50)
print(numbers)
```

```python
numbers = [10, 20, 40, 50]
numbers.insert(2, 30)
print(numbers)
```

# 2. Tuple Maipulation

# Define tuple of 5 colors

```python
colors = ("Red", "Blue", "Green", "Yellow", "Black")
print(colors)
```

# Tuples Cannot be modified

```python
print("Tuples are immutable, so new items cannot be added directly.")
```

# covert tuple to list

```python
colors = ("Red", "Blue", "Green", "Yellow", "Black")
color_list = list(colors)
print(color_list)
```

# Add new color

```python
colors = ("Red", "Blue", "Green", "Yellow", "Black")
color_list = list(colors)
color_list.append("White")
print(color_list)
```

# Convert back to tuple

```python
colors = tuple(color_list)
print(colors)
```
